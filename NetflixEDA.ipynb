# **Project Summary -**

The project **Netflix Movies and TV Shows** focuses on exploring Netflix's content catalogue and applying unsupervised machine learning techniques to identify similarities between movies and television shows. The primary objective was to understand the composition of Netflix's catalogue, discover patterns in its content distribution, and group similar titles based on their genres and descriptions.

The project began with an exploratory analysis of the Netflix Movies and TV Shows dataset, which contained 7,787 records and 12 original columns. Data wrangling was performed to handle missing values, remove unnecessary whitespace, convert date columns into appropriate formats, and extract additional features such as content duration, duration type, year added, and month added. Missing categorical values were replaced with 'Unknown', while unavailable date information was retained as missing rather than artificially filled.

Exploratory Data Analysis (EDA) was conducted using different visualizations to understand the catalogue. The analysis revealed that Movies accounted for approximately 69.1% of the dataset, while TV Shows represented 30.9%. The United States was the largest content-producing country, followed by India and the United Kingdom. Movies dominated the catalogue across most major countries, although South Korea and Japan showed a comparatively higher representation of TV Shows. The analysis also highlighted the growth of Netflix's catalogue over time, with 2019 recording the highest number of content additions. Content ratings indicated a strong representation of TV-MA and TV-14 titles. Additional visualizations examined genres, movie durations, TV Show seasons, directors, and the relationship between original release years and Netflix addition years.

Three statistical hypothesis tests were conducted to validate important observations. A one-sample Binomial Test confirmed that the proportion of Movies was significantly greater than 50%. A Welch's independent two-sample t-test identified a statistically significant difference in average movie durations between titles released before 2015 and those released from 2015 onwards. A Chi-Square Test of Independence also established a statistically significant association between country and content type. These tests provided statistical support for the observed patterns, while their results were interpreted as associations rather than evidence of causation.

For machine learning, the project focused on textual data preprocessing and feature extraction. Genres and descriptions were combined to create a text-based representation of each title. Text cleaning techniques included lowercasing, removing punctuation, URLs, numbers and stopwords, tokenization, and lemmatization. TF-IDF (Term Frequency–Inverse Document Frequency) was used to convert textual information into numerical features, while Truncated SVD reduced the dimensionality of the resulting feature matrix.

Three unsupervised learning algorithms were implemented: **K-Means Clustering, Agglomerative Hierarchical Clustering, and Gaussian Mixture Model (GMM)**. Hyperparameter tuning was performed using a manual grid-based search, and the Silhouette Score was used as the primary clustering evaluation metric. The optimized K-Means model achieved a Silhouette Score of 0.1184, while the optimized GMM achieved 0.1045. Although Agglomerative Clustering recorded a higher score of 0.2164, its resulting clusters were extremely imbalanced, with 7,785 titles in one cluster and only two in the other. Therefore, K-Means was selected as the final exploratory model because it provided a more practical clustering solution among the evaluated models.

Feature interpretation was performed by examining the average TF-IDF weights of terms within each cluster to understand the common themes represented by the groups. This helped provide insight into the textual characteristics of the clustered titles.

Overall, the project demonstrates how exploratory data analysis, statistical testing, natural language processing, and unsupervised machine learning can be combined to understand a large entertainment catalogue. The findings could support content categorization, catalogue analysis, and content discovery. However, the relatively low Silhouette Score indicates that the clusters have limited separation, and further feature engineering and model experimentation would be necessary to develop stronger content segments. Since the dataset does not contain individual viewing histories or user preferences, the resulting model should be considered a foundation for content-based exploration rather than a personalized recommendation system.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python

# Creating a copy of the original dataset

netflix_df = titles_df.copy()

print("Working copy created successfully!")
print("Dataset Shape:", netflix_df.shape)
````

````python

# Checking missing values in the working dataset

print("Missing Values Before Treatment:\n")
print(netflix_df.isnull().sum())
````

````python
columns_to_fill = ['director', 'cast', 'country', 'rating']

for column in columns_to_fill:
    netflix_df[column] = netflix_df[column].fillna('Unknown')

print("Missing values handled successfully!")
````

````python

# Checking missing values after treatment

print("Missing Values After Treatment:\n")
print(netflix_df.isnull().sum())
````

````python

# Converting date_added into datetime format

netflix_df['date_added'] = pd.to_datetime(
    netflix_df['date_added'].str.strip(),
    errors='coerce'
)

print("Date column converted successfully!")
print("Data Type:", netflix_df['date_added'].dtype)
````

````python

# Standardizing text columns

text_columns = netflix_df.select_dtypes(include='object').columns

for column in text_columns:
    netflix_df[column] = netflix_df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text columns standardized successfully!")
````

````python

# Extracting duration value

netflix_df['duration_int'] = pd.to_numeric(
    netflix_df['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Extracting duration unit

netflix_df['duration_type'] = netflix_df['duration'].str.extract(
    r'([a-zA-Z]+)'
)[0]

print("Duration information extracted successfully!")
````

````python

# Viewing extracted duration information

display(
    netflix_df[['title', 'type', 'duration',
                'duration_int', 'duration_type']].head(10)
)
````

````python

# Convert date_added to datetime format
netflix_df['date_added'] = pd.to_datetime(
    netflix_df['date_added'].astype(str).str.strip(),
    errors='coerce'
)

# Extracting year and month from date_added
netflix_df['year_added'] = netflix_df['date_added'].dt.year.astype('Int64')
netflix_df['month_added'] = netflix_df['date_added'].dt.month.astype('Int64')

print("Date features created successfully!")

# Verify the extracted columns
print(netflix_df[['date_added', 'year_added', 'month_added']].head())

# Check datatypes
print("\nColumn Datatypes:")
print(netflix_df[['date_added', 'year_added', 'month_added']].dtypes)
````

#### Chart - 1

````python

# Chart 1: Movies vs TV Shows Distribution

# Calculate content distribution
content_counts = netflix_df['type'].value_counts()
total_titles = content_counts.sum()

# Netflix-inspired styling
colors = ['#E50914', '#A8A8A8']

fig, ax = plt.subplots(figsize=(10, 7), facecolor='#141414')
ax.set_facecolor('#141414')

# Create donut chart
wedges, texts, autotexts = ax.pie(
    content_counts,
    labels=None,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct='%1.1f%%',
    pctdistance=0.78,
    wedgeprops={
        'width': 0.38,
        'edgecolor': '#141414',
        'linewidth': 4
    },
    textprops={
        'color': 'white',
        'fontsize': 13,
        'fontweight': 'bold'
    }
)

# Centre text
ax.text(
    0, 0.08,
    f'{total_titles:,}',
    ha='center',
    va='center',
    fontsize=27,
    fontweight='bold',
    color='white'
)

ax.text(
    0, -0.16,
    'TOTAL TITLES',
    ha='center',
    va='center',
    fontsize=10,
    color='#B3B3B3',
    fontweight='bold'
)

# Chart title and subtitle
ax.set_title(
    'NETFLIX CONTENT DISTRIBUTION',
    fontsize=17,
    fontweight='bold',
    color='white',
    pad=25
)

ax.text(
    0.5, 0.91,
    'Movies vs TV Shows',
    transform=ax.transAxes,
    ha='center',
    fontsize=11,
    color='#B3B3B3'
)

# Custom legend
legend_labels = [
    f'{category}: {count:,} ({count / total_titles * 100:.1f}%)'
    for category, count in content_counts.items()
]

legend = ax.legend(
    wedges,
    legend_labels,
    title='CONTENT TYPE',
    loc='lower center',
    bbox_to_anchor=(0.5, -0.12),
    ncol=2,
    frameon=False,
    fontsize=11,
    title_fontsize=10
)

plt.setp(legend.get_texts(), color='white')
legend.get_title().set_color('#B3B3B3')

ax.axis('equal')
plt.tight_layout()
plt.show()
````

#### Chart - 2

````python

# Chart 2: Top 10 Content Genres on Netflix
# Excluding broad categories

excluded_genres = [
    'International Movies',
    'International TV Shows',
    'Independent Movies'
]

# Split multiple genres into individual entries and count them
genre_counts = (
    netflix_df['listed_in']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
)

# Exclude selected broad categories
genre_counts = genre_counts[~genre_counts.isin(excluded_genres)]

# Get top 10 genres
genre_counts = genre_counts.value_counts().head(10)

# Netflix-inspired styling
fig, ax = plt.subplots(figsize=(11, 7), facecolor='#141414')
ax.set_facecolor('#141414')

bars = ax.barh(
    genre_counts.index[::-1],
    genre_counts.values[::-1],
    color=['#A8A8A8'] * 9 + ['#E50914'],
    height=0.65
)

# Add count labels
for bar in bars:
    width = bar.get_width()
    ax.text(
        width + 15,
        bar.get_y() + bar.get_height() / 2,
        f'{int(width):,}',
        va='center',
        color='white',
        fontsize=10,
        fontweight='bold'
    )

# Title and subtitle
ax.set_title(
    'TOP 10 CONTENT GENRES ON NETFLIX',
    fontsize=17,
    fontweight='bold',
    color='white',
    pad=22
)

ax.text(
    0.5, 1.01,
    'Excluding broad international and independent categories',
    transform=ax.transAxes,
    ha='center',
    fontsize=10,
    color='#A8A8A8'
)

# Axis styling
ax.set_xlabel('NUMBER OF TITLES', fontsize=10, color='#B3B3B3', labelpad=12)
ax.set_ylabel('GENRE', fontsize=10, color='#B3B3B3', labelpad=12)

ax.tick_params(axis='x', colors='#B3B3B3', labelsize=9)
ax.tick_params(axis='y', colors='white', labelsize=10)

ax.xaxis.grid(True, linestyle='--', alpha=0.2, color='#777777')
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.set_xlim(0, genre_counts.max() * 1.15)

plt.tight_layout()
plt.show()
````

````python

# Chart 4: Movies vs TV Shows by Top 10 Countries

# Create a separate dataset by splitting multiple countries
country_type_df = netflix_df[['country', 'type']].copy()

country_type_df['country'] = country_type_df['country'].str.split(',')
country_type_df = country_type_df.explode('country')

country_type_df['country'] = country_type_df['country'].str.strip()

# Exclude unknown and empty country values
country_type_df = country_type_df[
    (country_type_df['country'] != '') &
    (country_type_df['country'].str.lower() != 'unknown')
]

# Identify the top 10 countries based on total title associations
top_10_countries = (
    country_type_df['country']
    .value_counts()
    .head(10)
    .index
)

# Count Movies and TV Shows separately for each country
country_type_counts = (
    country_type_df[
        country_type_df['country'].isin(top_10_countries)
    ]
    .groupby(['country', 'type'])
    .size()
    .unstack(fill_value=0)
)

# Sort countries by total number of titles
country_type_counts['Total'] = country_type_counts.sum(axis=1)
country_type_counts = country_type_counts.sort_values(
    'Total', ascending=False
)

# Remove total column for plotting
country_type_counts = country_type_counts.drop(columns='Total')

# Netflix-inspired styling
fig, ax = plt.subplots(figsize=(12, 8), facecolor='#141414')
ax.set_facecolor('#141414')

# Plot stacked horizontal bars
plot_data = country_type_counts.iloc[::-1]

movies = plot_data['Movie']
tv_shows = plot_data['TV Show']

bars_movies = ax.barh(
    plot_data.index,
    movies,
    color='#E50914',
    label='Movies',
    height=0.65
)

bars_tv = ax.barh(
    plot_data.index,
    tv_shows,
    left=movies,
    color='#A8A8A8',
    label='TV Shows',
    height=0.65
)

# Add segment labels where there is enough space
for bar in bars_movies:
    width = bar.get_width()
    if width > 80:
        ax.text(
            width / 2,
            bar.get_y() + bar.get_height() / 2,
            f'{int(width):,}',
            ha='center',
            va='center',
            color='white',
            fontsize=9,
            fontweight='bold'
        )

for bar in bars_tv:
    width = bar.get_width()
    if width > 80:
        ax.text(
            bar.get_x() + width / 2,
            bar.get_y() + bar.get_height() / 2,
            f'{int(width):,}',
            ha='center',
            va='center',
            color='#141414',
            fontsize=9,
            fontweight='bold'
        )

# Add total counts at the end of each bar
for i, country in enumerate(plot_data.index):
    total = plot_data.loc[country].sum()
    ax.text(
        total + 20,
        i,
        f'{total:,}',
        va='center',
        color='white',
        fontsize=9,
        fontweight='bold'
    )

# Title and subtitle
ax.set_title(
    'MOVIES VS TV SHOWS BY TOP 10 COUNTRIES',
    fontsize=17,
    fontweight='bold',
    color='white',
    pad=24
)

ax.text(
    0.5, 1.01,
    'Country-wise composition of Netflix’s content catalogue',
    transform=ax.transAxes,
    ha='center',
    fontsize=10,
    color='#A8A8A8'
)

# Axis styling
ax.set_xlabel('NUMBER OF TITLES', fontsize=10, color='#B3B3B3', labelpad=12)
ax.set_ylabel('COUNTRY', fontsize=10, color='#B3B3B3', labelpad=12)

ax.tick_params(axis='x', colors='#B3B3B3', labelsize=9)
ax.tick_params(axis='y', colors='white', labelsize=10)

ax.xaxis.grid(True, linestyle='--', alpha=0.2, color='#777777')
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

# Legend
legend = ax.legend(
    loc='lower right',
    frameon=False,
    fontsize=10,
    title='CONTENT TYPE',
    title_fontsize=9
)

plt.setp(legend.get_texts(), color='white')
legend.get_title().set_color('#B3B3B3')

ax.set_xlim(0, plot_data.sum(axis=1).max() * 1.18)
print(plot_data)
plt.tight_layout()
plt.show()
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

**Null Hypothesis (H₀):** The proportion of Movies in the Netflix catalogue is equal to or less than 50%. In other words, Movies do not make up a statistically significant majority of the catalogue.

**Alternative Hypothesis (H₁):** The proportion of Movies in the Netflix catalogue is greater than 50%, indicating that Movies make up a statistically significant majority of the catalogue.

#### 2. Perform an appropriate statistical test.

````python

from scipy.stats import binomtest

# Count Movies and TV Shows
movie_count = (netflix_df['type'] == 'Movie').sum()
tv_count = (netflix_df['type'] == 'TV Show').sum()

total_count = movie_count + tv_count

# Perform one-sided Binomial Test
result = binomtest(
    k=movie_count,
    n=total_count,
    p=0.5,
    alternative='greater'
)

print("Movie Count:", movie_count)
print("TV Show Count:", tv_count)
print("Total Titles:", total_count)
print("Movie Percentage:", round((movie_count / total_count) * 100, 2), "%")
print("P-Value:", result.pvalue)
````

##### Which statistical test have you done to obtain P-Value?

The significance level (α) is 0.05.

Since the calculated p-value (2.834 × 10⁻²⁵⁴) is significantly smaller than 0.05, we reject the null hypothesis (H₀).

There is sufficient statistical evidence to conclude that the proportion of Movies in the Netflix catalogue is significantly greater than 50%.

The dataset contains 5,377 Movies (69.05%) compared to 2,410 TV Shows (30.95%). Therefore, the research hypothesis is statistically supported.

##### Why did you choose the specific statistical test?

I selected the **One-Sample Binomial Test** because the variable `type` consists of two categories: Movies and TV Shows. The objective of this hypothesis was to determine whether the proportion of Movies in the Netflix catalogue is significantly greater than 50%.

The Binomial Test is appropriate because it compares the observed proportion of a categorical outcome against a predefined theoretical proportion. Since our alternative hypothesis states that the Movie proportion is greater than 50%, I used a one-tailed test with a significance level of 5% (α = 0.05).

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

The average duration of Movies released in recent years is significantly different from the average duration of Movies released in earlier years.
**Null Hypothesis (H₀):** There is no significant difference in the average duration of Movies released in earlier years and Movies released in recent years.

**Alternative Hypothesis (H₁):** There is a significant difference in the average duration of Movies released in earlier years and Movies released in recent years.

#### 2. Perform an appropriate statistical test.

````python

from scipy.stats import ttest_ind

# Select Movies only
movie_df = netflix_df[netflix_df['type'] == 'Movie'].copy()

# Divide Movies into earlier and recent release periods
earlier_movies = movie_df[
    movie_df['release_year'] <= 2014
]['duration_int'].dropna()

recent_movies = movie_df[
    movie_df['release_year'] >= 2015
]['duration_int'].dropna()

# Perform Welch's Independent Two-Sample t-test
result = ttest_ind(
    earlier_movies,
    recent_movies,
    equal_var=False,
    alternative='two-sided'
)

print("Earlier Movies (up to 2014):", len(earlier_movies))
print("Recent Movies (2015 onwards):", len(recent_movies))

print("\nEarlier Movies - Average Duration:",
      round(earlier_movies.mean(), 2), "minutes")

print("Recent Movies - Average Duration:",
      round(recent_movies.mean(), 2), "minutes")

print("\nT-statistic:", result.statistic)
print("P-Value:", result.pvalue)

# Interpret the result
alpha = 0.05

if result.pvalue < alpha:
    print("\nConclusion: Reject the null hypothesis.")
    print("There is a statistically significant difference in average movie durations.")
else:
    print("\nConclusion: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of a significant difference in average movie durations.")
````

##### Which statistical test have you done to obtain P-Value?

I performed **Welch's Independent Two-Sample t-test (two-tailed)** to obtain the p-value. The test compares the average durations of Movies released up to 2014 with those released from 2015 onwards.

##### Why did you choose the specific statistical test?

I selected Welch's Independent Two-Sample t-test because the objective is to compare the average movie durations of two independent groups: earlier releases (up to 2014) and recent releases (2015 onwards).

This test is appropriate for comparing the means of two independent samples without assuming that their population variances are equal. I chose a two-tailed test because the objective is to determine whether there is a significant difference in average durations, irrespective of which group has the higher average.

Additionally, the dataset contains a sufficiently large number of Movies in both groups, making the test reasonably robust to deviations from normality. The significance level was set at 5% (α = 0.05).

#### 2. Perform an appropriate statistical test.

````python

from scipy.stats import chi2_contingency

# Create a copy of the required columns
country_df = netflix_df[['country', 'type']].copy()

# Split multiple countries into individual rows
country_df['country'] = country_df['country'].str.split(',')

country_df = country_df.explode('country')

# Remove extra spaces
country_df['country'] = country_df['country'].str.strip()

# Exclude missing and Unknown country values
country_df = country_df[
    country_df['country'].notna() &
    (country_df['country'] != '') &
    (country_df['country'].str.lower() != 'unknown')
]

# Create contingency table
contingency_table = pd.crosstab(
    country_df['country'],
    country_df['type']
)

# Perform Chi-Square Test of Independence
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("Degrees of Freedom:", dof)
print("P-Value:", p_value)

# Interpret the result
alpha = 0.05

if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis.")
    print("There is a statistically significant association between country and content type.")
else:
    print("\nConclusion: Fail to reject the null hypothesis.")
    print("There is insufficient evidence of an association between country and content type.")
````

#### 1. Expand Contraction

````python

# Install and import NLP libraries
!pip install contractions nltk -q

import re
import contractions
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

# Combine description and genre information
netflix_df['text_data'] = (
    netflix_df['listed_in'].fillna('') + ' ' +
    netflix_df['description'].fillna('')
)

print("Text data prepared successfully!")
print(netflix_df[['title', 'text_data']].head())


# Expand contractions
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: contractions.fix(text)
)

print("Contractions expanded successfully!")
print(netflix_df['text_data'].head())
````

````python

# Convert text to lowercase
netflix_df['text_data'] = netflix_df['text_data'].str.lower()

print("Lower casing completed!")
print(netflix_df['text_data'].head())
````

````python

# Remove punctuation
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: re.sub(r'[^\w\s]', ' ', text)
)

print("Punctuation removal completed!")
print(netflix_df['text_data'].head())
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python

# Remove URLs
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: re.sub(r'https?://\S+|www\.\S+', ' ', text)
)

# Remove tokens containing digits
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: re.sub(r'\b(?=\w*\d)\w+\b', ' ', text)
)

# Remove standalone numbers
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: re.sub(r'\b\d+\b', ' ', text)
)

print("URLs and digit-containing tokens removed!")
print(netflix_df['text_data'].head())
````

````python

# Load English stopwords
stop_words = set(stopwords.words('english'))

# Remove stopwords
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: ' '.join(
        word for word in text.split()
        if word not in stop_words
    )
)

# Remove extra white spaces
netflix_df['text_data'] = netflix_df['text_data'].apply(
    lambda text: re.sub(r'\s+', ' ', text).strip()
)

print("Stopword and whitespace removal completed!")
print(netflix_df['text_data'].head())
````

I selected lemmatization over stemming because it generally produces meaningful dictionary words rather than simply removing word endings. This is useful for identifying similarities between Netflix titles based on their descriptions and genres.

````python

# ML MODEL 1 - K-MEANS CLUSTERING

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt
import numpy as np

# Use dimensionality-reduced features if TruncatedSVD has been applied.
# Otherwise, use the TF-IDF matrix directly.
X = reduced_features if 'reduced_features' in globals() else tfidf_matrix

# Initialize the K-Means model
kmeans_model = KMeans(
    n_clusters=5,
    init='k-means++',
    n_init=10,
    max_iter=300,
    random_state=42
)

# Fit the algorithm
kmeans_model.fit(X)

print("K-Means model fitted successfully!")




# Predict cluster assignments
kmeans_labels = kmeans_model.predict(X)

# Store cluster labels in the dataset
netflix_df['KMeans_Cluster'] = kmeans_labels

# Display cluster distribution
print("Cluster Distribution:")
print(netflix_df['KMeans_Cluster'].value_counts().sort_index())

# Calculate Silhouette Score
kmeans_silhouette = silhouette_score(
    X,
    kmeans_labels,
    sample_size=min(3000, X.shape[0]),
    random_state=42
)

print("\nK-Means Silhouette Score:", round(kmeans_silhouette, 4))



# Visualize K-Means evaluation metric

plt.figure(figsize=(8, 5))
plt.bar(
    ['K-Means'],
    [kmeans_silhouette],
    color='#E50914',
    width=0.4
)

plt.title(
    'K-MEANS CLUSTERING - SILHOUETTE SCORE',
    color='white',
    fontsize=14,
    fontweight='bold'
)
plt.ylabel('Silhouette Score', color='white')
plt.ylim(-1, 1)

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

for i, score in enumerate([kmeans_silhouette]):
    plt.text(i, score + 0.04, f'{score:.4f}',
             ha='center', color='white', fontweight='bold')

plt.show()

````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

K-Means is an unsupervised clustering algorithm that divides data points into a predefined number of clusters based on their similarity. It assigns each data point to the nearest cluster centroid and iteratively updates the centroids to minimize the within-cluster sum of squared distances.

In this project, K-Means was applied to the TF-IDF feature representations of Netflix titles. The algorithm groups titles based on similarities in their genres and descriptions.

The Silhouette Score was used to evaluate clustering performance. It measures how similar a data point is to its own cluster compared with other clusters. A higher score indicates better-defined and more separated clusters.

#### 2. Cross- Validation & Hyperparameter Tuning

````python

# CROSS-VALIDATION & HYPERPARAMETER TUNING

from sklearn.model_selection import ParameterGrid
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import pandas as pd
import matplotlib.pyplot as plt

# Define hyperparameter grid
param_grid = {
    'n_clusters': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'init': ['k-means++', 'random'],
    'n_init': [10, 20],
    'max_iter': [300]
}

best_score = -1
best_params = None
tuning_results = []

# Evaluate different hyperparameter combinations
for params in ParameterGrid(param_grid):

    model = KMeans(
        **params,
        random_state=42
    )

    labels = model.fit_predict(X)

    score = silhouette_score(
        X,
        labels,
        sample_size=min(3000, X.shape[0]),
        random_state=42
    )

    tuning_results.append({
        'Parameters': str(params),
        'Silhouette Score': score
    })

    if score > best_score:
        best_score = score
        best_params = params

# Display best configuration
print("Best Hyperparameters:", best_params)
print("Best Silhouette Score:", round(best_score, 4))

# Fit the optimized model
optimized_kmeans = KMeans(
    **best_params,
    random_state=42
)

optimized_labels = optimized_kmeans.fit_predict(X)

# Store optimized cluster assignments
netflix_df['KMeans_Optimized_Cluster'] = optimized_labels

# Calculate final evaluation score
optimized_silhouette = silhouette_score(
    X,
    optimized_labels,
    sample_size=min(3000, X.shape[0]),
    random_state=42
)

print("Optimized Silhouette Score:", round(optimized_silhouette, 4))

# Compare baseline and optimized models
comparison = pd.DataFrame({
    'Model': ['Baseline K-Means', 'Optimized K-Means'],
    'Silhouette Score': [
        kmeans_silhouette,
        optimized_silhouette
    ]
})

print("\nModel Performance Comparison:")
print(comparison)

# Visualize evaluation metric comparison
plt.figure(figsize=(9, 5))

bars = plt.bar(
    comparison['Model'],
    comparison['Silhouette Score'],
    color=['#A8A8A8', '#E50914'],
    width=0.5
)

plt.title(
    'K-MEANS: BASELINE VS OPTIMIZED PERFORMANCE',
    color='white',
    fontsize=13,
    fontweight='bold'
)
plt.ylabel('Silhouette Score', color='white')

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.02,
        f'{height:.4f}',
        ha='center',
        color='white',
        fontweight='bold'
    )

plt.show()
````

````python

# ============================================================
# ML MODEL 2 - AGGLOMERATIVE HIERARCHICAL CLUSTERING
# ============================================================

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
from sklearn.model_selection import ParameterGrid
from sklearn.decomposition import TruncatedSVD

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PREPARE FEATURES
# ------------------------------------------------------------

# Use reduced features if already generated; otherwise apply SVD
if 'reduced_features' in globals():
    X_agg = reduced_features
else:
    svd_agg = TruncatedSVD(n_components=100, random_state=42)
    X_agg = svd_agg.fit_transform(tfidf_matrix)

print("Feature Matrix Shape:", X_agg.shape)


# ------------------------------------------------------------
# 2. BASELINE MODEL - FIT THE ALGORITHM
# ------------------------------------------------------------

agg_model = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

agg_labels = agg_model.fit_predict(X_agg)

netflix_df['Agglomerative_Cluster'] = agg_labels

agg_silhouette = silhouette_score(
    X_agg,
    agg_labels,
    sample_size=min(3000, X_agg.shape[0]),
    random_state=42
)

print("\nBASELINE AGGLOMERATIVE CLUSTERING")
print("Silhouette Score:", round(agg_silhouette, 4))

print("\nCluster Distribution:")
print(pd.Series(agg_labels).value_counts().sort_index())


# ------------------------------------------------------------
# 3. BASELINE EVALUATION VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    ['Agglomerative Clustering'],
    [agg_silhouette],
    color='#E50914',
    width=0.4
)

plt.title(
    'AGGLOMERATIVE CLUSTERING - SILHOUETTE SCORE',
    color='white',
    fontsize=13,
    fontweight='bold'
)

plt.ylabel('Silhouette Score', color='white')
plt.ylim(-1, 1)

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

plt.text(
    0,
    agg_silhouette + 0.04,
    f'{agg_silhouette:.4f}',
    ha='center',
    color='white',
    fontweight='bold'
)

plt.show()


# ------------------------------------------------------------
# 4. HYPERPARAMETER TUNING - SAMPLE-BASED GRID SEARCH
# ------------------------------------------------------------

# Select a representative sample to reduce computation time
rng = np.random.RandomState(42)

sample_size = min(2000, X_agg.shape[0])

sample_indices = rng.choice(
    X_agg.shape[0],
    size=sample_size,
    replace=False
)

X_sample = X_agg[sample_indices]

print("\nTUNING SAMPLE SHAPE:", X_sample.shape)

# Define a manageable parameter grid
param_grid = {
    'n_clusters': [2, 3, 4, 5, 6],
    'linkage': ['ward', 'average']
}

best_agg_score = -1
best_agg_params = None
agg_tuning_results = []

for params in ParameterGrid(param_grid):

    model = AgglomerativeClustering(**params)

    labels = model.fit_predict(X_sample)

    # Skip configurations producing fewer than 2 clusters
    if len(np.unique(labels)) < 2:
        print("Skipping invalid configuration:", params)
        continue

    score = silhouette_score(
        X_sample,
        labels,
        sample_size=min(1500, sample_size),
        random_state=42
    )

    agg_tuning_results.append({
        'Parameters': str(params),
        'Silhouette Score': score
    })

    if score > best_agg_score:
        best_agg_score = score
        best_agg_params = params

# Display all valid tuning results
print("\nHYPERPARAMETER TUNING RESULTS")
print(pd.DataFrame(agg_tuning_results))

if best_agg_params is None:
    raise ValueError("No valid clustering configuration was found.")

print("\nBest Hyperparameters:", best_agg_params)
print("Best Tuning Silhouette Score:", round(best_agg_score, 4))

# ------------------------------------------------------------
# 5. OPTIMIZED MODEL - FIT AND PREDICT
# ------------------------------------------------------------

optimized_agg = AgglomerativeClustering(
    **best_agg_params
)

optimized_agg_labels = optimized_agg.fit_predict(X_agg)

netflix_df['Agglomerative_Optimized_Cluster'] = optimized_agg_labels

print("\nOPTIMIZED CLUSTER DISTRIBUTION:")
print(
    pd.Series(optimized_agg_labels)
    .value_counts()
    .sort_index()
)

# Create a stratified evaluation sample
def get_stratified_sample(X, labels, max_per_cluster=1500, random_state=42):

    rng = np.random.RandomState(random_state)
    selected_indices = []

    for cluster in np.unique(labels):

        cluster_indices = np.where(labels == cluster)[0]

        sample_count = min(max_per_cluster, len(cluster_indices))

        selected = rng.choice(
            cluster_indices,
            size=sample_count,
            replace=False
        )

        selected_indices.extend(selected)

    selected_indices = np.array(selected_indices)

    return X[selected_indices], labels[selected_indices]


# Evaluate optimized model
X_eval, y_eval = get_stratified_sample(
    X_agg,
    optimized_agg_labels
)

if len(np.unique(y_eval)) < 2:
    raise ValueError("Evaluation sample contains fewer than 2 clusters.")

optimized_agg_silhouette = silhouette_score(
    X_eval,
    y_eval
)

print("\nOPTIMIZED AGGLOMERATIVE CLUSTERING")
print("Evaluation Sample Shape:", X_eval.shape)
print("Optimized Silhouette Score:",
      round(optimized_agg_silhouette, 4))


# ------------------------------------------------------------
# 6. BASELINE VS OPTIMIZED MODEL COMPARISON
# ------------------------------------------------------------

# Re-evaluate baseline using the same stratified approach
X_baseline_eval, y_baseline_eval = get_stratified_sample(
    X_agg,
    agg_labels
)

baseline_agg_silhouette = silhouette_score(
    X_baseline_eval,
    y_baseline_eval
)

agg_comparison = pd.DataFrame({
    'Model': [
        'Baseline Agglomerative',
        'Optimized Agglomerative'
    ],
    'Silhouette Score': [
        baseline_agg_silhouette,
        optimized_agg_silhouette
    ]
})

print("\nMODEL PERFORMANCE COMPARISON")
print(agg_comparison)

agg_improvement = (
    optimized_agg_silhouette - baseline_agg_silhouette
)

print("\nScore Improvement:", round(agg_improvement, 4))


# ------------------------------------------------------------
# 7. UPDATED EVALUATION METRIC VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

bars = plt.bar(
    agg_comparison['Model'],
    agg_comparison['Silhouette Score'],
    color=['#A8A8A8', '#E50914'],
    width=0.5
)

plt.title(
    'AGGLOMERATIVE - BASELINE VS OPTIMIZED PERFORMANCE',
    color='white',
    fontsize=13,
    fontweight='bold'
)

plt.ylabel('Silhouette Score', color='white')

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.01,
        f'{height:.4f}',
        ha='center',
        color='white',
        fontweight='bold'
    )

plt.show()

print("\nAGGLOMERATIVE CLUSTERING COMPLETED SUCCESSFULLY!")
````

````python

# ============================================================
# ML MODEL 3 - GAUSSIAN MIXTURE MODEL (GMM)
# ============================================================

from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
from sklearn.model_selection import ParameterGrid
from sklearn.decomposition import TruncatedSVD

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. PREPARE FEATURES
# ------------------------------------------------------------

# Use reduced features if already generated; otherwise apply SVD
if 'reduced_features' in globals():
    X_gmm = reduced_features
else:
    svd_gmm = TruncatedSVD(n_components=100, random_state=42)
    X_gmm = svd_gmm.fit_transform(tfidf_matrix)

print("Feature Matrix Shape:", X_gmm.shape)


# ------------------------------------------------------------
# 2. BASELINE MODEL - FIT THE ALGORITHM
# ------------------------------------------------------------

gmm_model = GaussianMixture(
    n_components=2,
    covariance_type='diag',
    random_state=42,
    n_init=3,
    max_iter=200,
    reg_covar=1e-5
)

gmm_model.fit(X_gmm)

# Predict cluster assignments
gmm_labels = gmm_model.predict(X_gmm)

# Store cluster assignments
netflix_df['GMM_Cluster'] = gmm_labels

# Create a stratified evaluation sample
def get_gmm_sample(X, labels, max_per_cluster=1500, random_state=42):

    rng = np.random.RandomState(random_state)
    selected_indices = []

    for cluster in np.unique(labels):

        cluster_indices = np.where(labels == cluster)[0]

        sample_count = min(
            max_per_cluster,
            len(cluster_indices)
        )

        selected = rng.choice(
            cluster_indices,
            size=sample_count,
            replace=False
        )

        selected_indices.extend(selected)

    selected_indices = np.array(selected_indices)

    return X[selected_indices], labels[selected_indices]


X_gmm_eval, y_gmm_eval = get_gmm_sample(
    X_gmm,
    gmm_labels
)

if len(np.unique(y_gmm_eval)) < 2:
    raise ValueError("Baseline GMM produced fewer than 2 clusters.")

gmm_silhouette = silhouette_score(
    X_gmm_eval,
    y_gmm_eval
)

print("\nBASELINE GAUSSIAN MIXTURE MODEL")
print("Silhouette Score:", round(gmm_silhouette, 4))

print("\nCluster Distribution:")
print(pd.Series(gmm_labels).value_counts().sort_index())


# ------------------------------------------------------------
# 3. BASELINE EVALUATION VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    ['Gaussian Mixture Model'],
    [gmm_silhouette],
    color='#E50914',
    width=0.4
)

plt.title(
    'GAUSSIAN MIXTURE MODEL - SILHOUETTE SCORE',
    color='white',
    fontsize=13,
    fontweight='bold'
)

plt.ylabel('Silhouette Score', color='white')

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

plt.text(
    0,
    gmm_silhouette + 0.01,
    f'{gmm_silhouette:.4f}',
    ha='center',
    color='white',
    fontweight='bold'
)

plt.show()


# ------------------------------------------------------------
# 4. HYPERPARAMETER TUNING - GRID SEARCH
# ------------------------------------------------------------

# Use a representative sample to reduce computation time
rng = np.random.RandomState(42)

tuning_size = min(2000, X_gmm.shape[0])

tuning_indices = rng.choice(
    X_gmm.shape[0],
    size=tuning_size,
    replace=False
)

X_gmm_tuning = X_gmm[tuning_indices]

print("\nTUNING SAMPLE SHAPE:", X_gmm_tuning.shape)

# Define hyperparameter combinations
param_grid = {
    'n_components': [2, 3, 4, 5, 6],
    'covariance_type': ['diag', 'spherical']
}

best_gmm_score = -1
best_gmm_params = None
gmm_tuning_results = []

for params in ParameterGrid(param_grid):

    model = GaussianMixture(
        **params,
        random_state=42,
        n_init=2,
        max_iter=200,
        reg_covar=1e-5
    )

    model.fit(X_gmm_tuning)

    labels = model.predict(X_gmm_tuning)

    # Skip invalid configurations
    if len(np.unique(labels)) < 2:
        continue

    score = silhouette_score(
        X_gmm_tuning,
        labels,
        sample_size=min(1500, tuning_size),
        random_state=42
    )

    bic_score = model.bic(X_gmm_tuning)

    gmm_tuning_results.append({
        'Parameters': str(params),
        'Silhouette Score': score,
        'BIC Score': bic_score
    })

    if score > best_gmm_score:
        best_gmm_score = score
        best_gmm_params = params

# Display tuning results
gmm_results_df = pd.DataFrame(gmm_tuning_results)

print("\nGMM HYPERPARAMETER TUNING RESULTS")
print(gmm_results_df)

if best_gmm_params is None:
    raise ValueError("No valid GMM configuration was found.")

print("\nBest Hyperparameters:", best_gmm_params)
print("Best Tuning Silhouette Score:", round(best_gmm_score, 4))


# ------------------------------------------------------------
# 5. OPTIMIZED MODEL - FIT AND PREDICT
# ------------------------------------------------------------

optimized_gmm = GaussianMixture(
    **best_gmm_params,
    random_state=42,
    n_init=3,
    max_iter=200,
    reg_covar=1e-5
)

optimized_gmm.fit(X_gmm)

# Predict optimized cluster assignments
optimized_gmm_labels = optimized_gmm.predict(X_gmm)

netflix_df['GMM_Optimized_Cluster'] = optimized_gmm_labels

# Evaluate using a stratified sample
X_optimized_eval, y_optimized_eval = get_gmm_sample(
    X_gmm,
    optimized_gmm_labels
)

if len(np.unique(y_optimized_eval)) < 2:
    raise ValueError("Optimized GMM produced fewer than 2 clusters.")

optimized_gmm_silhouette = silhouette_score(
    X_optimized_eval,
    y_optimized_eval
)

print("\nOPTIMIZED GAUSSIAN MIXTURE MODEL")
print("Optimized Silhouette Score:",
      round(optimized_gmm_silhouette, 4))

print("\nOptimized Cluster Distribution:")
print(
    pd.Series(optimized_gmm_labels)
    .value_counts()
    .sort_index()
)


# ------------------------------------------------------------
# 6. BASELINE VS OPTIMIZED MODEL COMPARISON
# ------------------------------------------------------------

gmm_comparison = pd.DataFrame({
    'Model': [
        'Baseline GMM',
        'Optimized GMM'
    ],
    'Silhouette Score': [
        gmm_silhouette,
        optimized_gmm_silhouette
    ]
})

print("\nGMM MODEL PERFORMANCE COMPARISON")
print(gmm_comparison)

gmm_improvement = (
    optimized_gmm_silhouette - gmm_silhouette
)

print("\nScore Improvement:", round(gmm_improvement, 4))


# ------------------------------------------------------------
# 7. UPDATED EVALUATION METRIC VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

bars = plt.bar(
    gmm_comparison['Model'],
    gmm_comparison['Silhouette Score'],
    color=['#A8A8A8', '#E50914'],
    width=0.5
)

plt.title(
    'GMM - BASELINE VS OPTIMIZED PERFORMANCE',
    color='white',
    fontsize=13,
    fontweight='bold'
)

plt.ylabel('Silhouette Score', color='white')

plt.gca().set_facecolor('#141414')
plt.gcf().set_facecolor('#141414')
plt.tick_params(colors='white')
plt.grid(axis='y', alpha=0.2, color='grey')

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.01,
        f'{height:.4f}',
        ha='center',
        color='white',
        fontweight='bold'
    )

plt.show()

print("\nGAUSSIAN MIXTURE MODEL COMPLETED SUCCESSFULLY!")
````

````python

from sklearn.mixture import GaussianMixture
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import silhouette_score
import numpy as np
import pandas as pd

# Use the reduced feature matrix
X_gmm = reduced_features

# 1. Create a sample for hyperparameter tuning
np.random.seed(42)
sample_size = min(2000, X_gmm.shape[0])
sample_indices = np.random.choice(
    X_gmm.shape[0], sample_size, replace=False
)
X_tune = X_gmm[sample_indices]

print("Tuning Sample Shape:", X_tune.shape)

# 2. Define hyperparameter combinations
param_grid = {
    'n_components': [2, 3, 4, 5, 6],
    'covariance_type': ['diag', 'spherical']
}

# 3. Evaluate each combination
results = []

for params in ParameterGrid(param_grid):
    try:
        model = GaussianMixture(
            n_components=params['n_components'],
            covariance_type=params['covariance_type'],
            random_state=42,
            n_init=3,
            max_iter=200,
            reg_covar=1e-5
        )

        labels = model.fit_predict(X_tune)

        if len(np.unique(labels)) < 2:
            continue

        score = silhouette_score(X_tune, labels)

        results.append({
            'n_components': params['n_components'],
            'covariance_type': params['covariance_type'],
            'silhouette_score': score,
            'bic': model.bic(X_tune)
        })

    except Exception as e:
        print(f"Skipped {params}: {e}")

# 4. Display tuning results
results_df = pd.DataFrame(results).sort_values(
    by='silhouette_score', ascending=False
)

display(results_df)

# 5. Select the best hyperparameters
best_result = results_df.iloc[0]

best_params = {
    'n_components': int(best_result['n_components']),
    'covariance_type': best_result['covariance_type']
}

print("\nBest Hyperparameters:", best_params)
print("Best Tuning Silhouette Score:",
      round(best_result['silhouette_score'], 4))

# 6. Train optimized GMM on the complete dataset
best_gmm = GaussianMixture(
    n_components=best_params['n_components'],
    covariance_type=best_params['covariance_type'],
    random_state=42,
    n_init=3,
    max_iter=200,
    reg_covar=1e-5
)

optimized_gmm_labels = best_gmm.fit_predict(X_gmm)
netflix_df['GMM_Optimized_Cluster'] = optimized_gmm_labels

# 7. Evaluate optimized model using a stratified sample
def stratified_silhouette(X, labels, max_per_cluster=1500):
    rng = np.random.RandomState(42)
    selected_indices = []

    for cluster in np.unique(labels):
        cluster_indices = np.where(labels == cluster)[0]
        sample_indices = rng.choice(
            cluster_indices,
            min(len(cluster_indices), max_per_cluster),
            replace=False
        )
        selected_indices.extend(sample_indices)

    return silhouette_score(
        X[selected_indices],
        labels[selected_indices]
    )

optimized_score = stratified_silhouette(
    X_gmm, optimized_gmm_labels
)

print("\nOptimized GMM Silhouette Score:",
      round(optimized_score, 4))

print("\nOptimized Cluster Distribution:")
print(pd.Series(optimized_gmm_labels).value_counts().sort_index())
````
